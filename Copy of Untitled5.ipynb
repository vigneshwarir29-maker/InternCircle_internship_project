```python
%%html
<!DOCTYPE html><html lang="en"><head><meta charset="UTF-8"><meta name="viewport" content="width=device-width,initial-scale=1"><title>To-Do List</title>
<style>
body{font-family:system-ui,sans-serif;max-width:420px;margin:20px auto;padding:0 15px}
input{padding:8px;width:65%}button{padding:8px 12px;cursor:pointer}
ul{padding:0;list-style:none}
li{display:flex;justify-content:space-between;align-items:center;padding:8px;border-bottom:1px solid #ddd}
li span{cursor:pointer;flex:1}.done span{text-decoration:line-through;color:gray}
</style></head><body>
<h2>To-Do List</h2>
<input id="t" placeholder="New task"><button id="add">Add</button>
<ul id="l"></ul>
<script>
let tasks=JSON.parse(localStorage.getItem('tasks')||"[]");
const save=()=>localStorage.setItem('tasks',JSON.stringify(tasks));
function render(){
  l.innerHTML='';
    tasks.forEach((t,i)=>{
        const li=document.createElement('li');
            if(t.done)li.className='done';
                const s=document.createElement('span');s.textContent=t.text;
                    s.onclick=()=>{t.done=!t.done;save();render()};
                        const b=document.createElement('button');b.textContent='✕';
                            b.onclick=()=>{tasks.splice(i,1);save();render()};
                                li.append(s,b);l.appendChild(li);
                                  });
                                  }
                                  function addTask(){const v=t.value.trim();if(!v)return;tasks.push({text:v,done:false});t.value='';save();render()}
                                  add.onclick=addTask;
                                  t.addEventListener('keydown',e=>{if(e.key==='Enter')addTask()});
                                  render();
                                  </script></body></html>
```
